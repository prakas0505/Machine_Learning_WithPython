```python
# Importing prebuilt dataset
from sklearn.datasets import fetch_california_housing
```

```python
housing_data = fetch_california_housing(as_frame=True)
df_housing = housing_data.frame
```

```python
df_housing.head()
```

```python
# compute correlation
housing_corr = df_housing.corr()
```

```python
print(housing_corr)
```

```python
# Seaborn
sns.heatmap(housing_corr, annot=True, square=True, fmt= ".2f", cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
# Histogram( Distribution Plot) -- Understanding Spread

plt.figure()
plt.hist(df["age"], bins=20)
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")
plt.show()
```

```python
# Seaborn Histogram
plt.figure()
sns.histplot(df["age"], bins=20, kde=True)
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")
plt.show()
```

```python
# Box plot -- used in the outlier detection

plt.figure()
sns.boxplot(x=df["age"])
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")
plt.show()
```

```python
plt.figure()
sns.boxplot(df["income"], orient='v')
plt.xlabel("Income")
plt.ylabel("Frequency")
plt.title("Income Distribution")
plt.show()
```

```python
# violin plot
plt.figure()
sns.violinplot(data = df, x ='segment', y = 'monthly_usage')
# plt.xlabel("Age")
# plt.ylabel("Frequency")
# plt.title("Age Distribution")
plt.show()
```

```python
# Scatter plot ( Relationship )
plt.figure()
sns.scatterplot(x=df["age"], y=df["monthly_usage"])
plt.xlabel("Age")
plt.ylabel("Monthly Usage")
plt.title("Age vs Monthly Usage")
plt.show()
```

```python
plt.figure()
plt.scatter(df["income"], df["purchase_amount"], c=df["age"], cmap='viridis')
plt.xlabel("Income")
plt.ylabel("Purchase Amount")
plt.title("Income vs Purchase Amount")
plt.legend()
plt.show()
```

```python
plt.figure()
sns.scatterplot(x="income", y = "purchase_amount", data=df, hue="age", palette="viridis", alpha =0.8)
plt.xlabel("Income")
plt.ylabel("Purchase Amount")
plt.title("Income vs Purchase Amount")
```

```python
# Line plot
plt.figure()
sns.lineplot(x=df["age"], y=df["monthly_usage"])
plt.xlabel("Age")
plt.ylabel("Monthly Usage")
plt.title("Age vs Monthly Usage")
plt.show()
```

```python
# Bar plot and count pllot

sns.countplot(x=df["segment"], hue=df['churn'], color='red')
plt.xlabel("Segment")
plt.ylabel("Count")
plt.title("Segment Distribution")
plt.show()
```

```python
# Barplot
sns.barplot(x = df["segment"], y = df["monthly_usage"], estimator=np.sum)
plt.xlabel("Segment")
plt.ylabel("Monthly Usage")
plt.title("Segment vs Monthly Usage")
plt.show()
```

```python
# parit plot
sns.pairplot(df)
plt.show()
```
