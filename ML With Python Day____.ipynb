# Data Visualization

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,10,100)
y = x**2

plt.plot(x, y, 'go')
plt.title("Title")
plt.xlabel("Ram")
plt.ylabel("Sham")
plt.show()
```

```python
plt.subplot(2,2,1)
plt.plot(x,y, 'go')
plt.title("Title")
plt.xlabel("Ram")
plt.ylabel("Sham")
plt.subplot(2,2,4)
plt.plot(x,y, 'go')
plt.title("2n Title")
plt.xlabel("X variable")
plt.ylabel("y variable")
```

```python
fig = plt.figure(figsize=(8,6))
ax1 = fig.add_axes([0.1,0.1,0.4,0.3])
ax2 = fig.add_axes([0.5,0.9,0.4,0.3])
ax1.plot(x,y)
ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.set_title('title')
ax2.plot(y,x)
ax2.set_xlabel('y')
ax2.set_ylabel('x')
ax2.set_title('title')
```
