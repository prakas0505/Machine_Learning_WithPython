# Pandas

```python
import pandas as pd
```

```python
# Create a Series
data = [10, 20, 30, 40]
s = pd.Series(data)
print(s)
```

```python
s = pd.Series(data, index=['a', 'b', 'c', 'e'])
print(s)
```

```python
# create a DataFrame

data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, 35],
    "City": ["New York", "San Francisco", "Los Angeles"]
}

df = pd.DataFrame(data)

print(df)
```
